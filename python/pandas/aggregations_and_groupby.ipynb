```python
import pandas as pd 
from faker import Faker 
import numpy as np 

fake = Faker()
np.random.seed(42)
cities = ['Rome', 'Milan', 'Florence', 'Naples']

n = 100
data = {
    'Name': [f"{fake.first_name()} {fake.last_name()}" for _ in range(n)],
    'Age' : np.random.randint(20, 60, n),
    'City': np.random.choice(cities, n),
    'Score' : np.random.randint(0,101, n)
}

df = pd.DataFrame(data)
df
```

```python
grouped = df.groupby('City')
grouped['Score'].mean()
```

```python
df.groupby('City')['Score'].agg(['mean','max','count'])
```

```python
print(df['Score'].mean())
grouped = df.groupby('City')
print(grouped['Score'].mean())

print(grouped['Name'].count())

print(grouped['Score'].max())

df.groupby('City').agg({
    'Score' : ['mean', 'max', 'count'],
    'Age'   : ['max', 'count']
})
```
