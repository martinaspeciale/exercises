```python
# complete exercise 
import pandas as pd
import numpy as np

np.random.seed(42)
n = 50

df = pd.DataFrame({
    'Date': np.random.choice(pd.date_range(start='2023-01-01', end='2024-01-01', freq='M'), n),
    'Value': np.random.randint(0, 101, n)
})

df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values('Date')
df = df.set_index('Date')

# Resample mensile
'''
resample() in pandas funziona quasi come un groupby(),
ma sul tempo
 --> un groupby temporale che permette di aggregare dati
     su una nuova frequenza temporale 
 --> è necessario avere un indice temporale (DatetimeIndex)
il nome "resample" perchè 
 --> prendi una serie temporale e ricampiona (riorganizza) i dati 
     su un'altra frequenza temporale 
     e.g. : se hai dati giornalieri, puoi aggregarli mensilmente,
            settimanalmente, annualmente etc.
 --> in pratica: ristruttura l'asse temporale creando intervalli regolari
     ed applica l'aggregazione scelta entro ogni intervallo
            
    groupby('column')
    resample('M') -> crea tutti i bucket temporali, anche se vuoti
         

'''
monthly_sum = df.resample('M')['Value'].sum()
print("Resample mensile:")
print(monthly_sum)

# Groupby per mese (numerico)
df['Month'] = df.index.month
grouped = df.groupby('Month')['Value'].sum().sort_index()
print("\nGroupby per mese:")
print(grouped)
```

```python
import pandas as pd

dates = pd.to_datetime(['2023-01-01', '2023-01-03', '2023-01-10'])
values = [10, 20, 30]

df = pd.DataFrame({'Date': dates, 'Value': values})
df = df.set_index('Date')

result = df.resample('D').sum()
print(result)
```
