```python
import pandas as pd

df1 = pd.DataFrame({
    'ID' : [1,2,3],
    'Name' : ['Alice', 'Bob', 'Charlie']
})

df2 = pd.DataFrame({
    'ID' : [4,5],
    'Name' : ['David', 'Emma']
})

df = pd.concat([df1, df2], axis=1, ignore_index = True)
df
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df = df.dropna() # elimina righe con NaN
df
```

```python
print(df)
df[2].fillna(df[2].mean())
```

```python
from faker import Faker 
import numpy as np 

np.random.seed(42)
Faker.seed(42)
fake = Faker()

n = 5

df_part1 = pd.DataFrame({
    'ID' : np.arange(n), 
    'Name' : [fake.first_name() for _ in range(n)],
    'Score' : np.random.randint(1, 101, n)
})

df_part2 = pd.DataFrame({
    'ID' : np.arange(n, 2*n), 
    'Name' : [fake.first_name() for _ in range(n)],
    'Score' : np.random.randint(1, 101, n)
})

df = pd.concat([df_part1, df_part2], ignore_index=True)
# df['ID'] = np.arange(len(df))
df
```

```python
# or we could use uuid for the IDs generation for df_part1, df_part2
import uuid

df_part1['ID'] = [str(uuid.uuid4()) for _ in range(n)]
df_part1
```

```python
random = np.random.randint(0,10,3)
print(random)
df.loc[random, 'Score'] = np.nan
df
```
