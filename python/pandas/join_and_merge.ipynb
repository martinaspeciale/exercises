```python
import pandas as pd
import numpy as np 
from faker import Faker 

np.random.seed(42)

df1 = pd.DataFrame({
    'ID' : [1,2,3],
    'Name': ['Alice', 'Bob', 'Charlie']
})

df2 = pd.DataFrame({
    'ID': [1,2,4],
    'Score': [80, 90, 70]
})

merged = pd.merge(df1, df2, on='ID', how='outer')
print(merged)

```

```python
pd.merge(df1, df2, left_index=True, right_index=True)
```

```python
fake = Faker()
n=10

cities = ['Roma', 'Milano', 'Bari', 'Firenze', 'Napoli', 'Pisa']

df_customers = pd.DataFrame({
    'CustomerID' : np.random.randint(0,10,n),
    'Name' : [f"{fake.first_name()} {fake.last_name()}" for _ in range(n)],
    'City' : np.random.choice(cities, n)
})

df_orders = pd.DataFrame({
    'OrderID' : np.arange(n),
    'CustomerID' : np.random.randint(0,10,n),
    'OrderAmount': np.random.randint(0,200,n)
})

print(pd.merge(df_customers, df_orders, on='CustomerID'))
print(pd.merge(df_customers, df_orders, how='left'))
print(pd.merge(df_customers, df_orders, how='outer'))
```
