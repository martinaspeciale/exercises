```python
import pandas as pd 
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt 

%matplotlib inline 
sns.set(style='ticks')
```

```python
url = 'https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Online_Retail/Online_Retail.csv'
df = pd.read_csv(url, encoding='latin1')

# default utf-8 didn't work -> non-UTF-8 bytes --> swith to latin1 encoding 
```

```python
print(f"Number of samples: {df.shape[0]}")
print(f"Number of features: {df.shape[1]} \n")
print(df.dtypes)
```

Create a histogram with the 10 countries that have the most 'Quantity' ordered except UK

```python
countries = df.groupby('Country')['Quantity'].agg(['sum','mean']).sort_values('sum', ascending=False).head(11)
countries
```

```python
countries.iloc[1:11]['sum'].plot(kind='bar')

plt.xlabel('Countries')
plt.ylabel('Quantity')
plt.title('10 Countries with most orders - UK excluded')
plt.xticks(rotation=45)

plt.show()
```
