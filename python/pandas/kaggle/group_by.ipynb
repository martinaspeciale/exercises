```python
import pandas as pd 
import numpy as np 

url = 'https://raw.githubusercontent.com/justmarkham/DAT8/master/data/drinks.csv'
```

```python
drinks = pd.read_csv(url)
drinks.head(3)
```

```python
for col in drinks.columns:
    print(f"* {col}")
```

Which continent drinks more beer on average?

```python
beer_per_continent = drinks.groupby('continent')['beer_servings'].mean()
print(beer_per_continent)
```
