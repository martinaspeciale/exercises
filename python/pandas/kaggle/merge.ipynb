```python
import pandas as pd 
import numpy as np 

url_1 = 'https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/05_Merge/Auto_MPG/cars1.csv'
url_2 = 'https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/05_Merge/Auto_MPG/cars2.csv'

cars1 = pd.read_csv(url_1)
cars2 = pd.read_csv(url_2)
```

It seems our first dataset has some unnamed blank columns, fix cars1

```python
cars1 = cars1.loc[:, :'car'] 
cars1.dtypes
```

What is the number of observations in each dataset?

```python
print(f"Observations in cars1: {cars1.shape[0]}")
print(f"Observations in cars2: {cars2.shape[0]}")
```

```python
def check_cols(one, two):
    return set(one.columns) == set(two.columns)

print(f"Equal Features Check: {'yes' if check_cols(cars1, cars2) else 'no'}")
```

Join cars1 and cars2 into a single DataFrame called cars

```python
cars = pd.concat([cars1,cars2])
cars.shape[0]
```
