Exercise 9: Concatenate two data frames using the following conditions

```python
GermanCars = {'Company': ['Ford', 'Mercedes', 'BMV', 'Audi'], 'Price': [23845, 171995, 135925 , 71400]}
japaneseCars = {'Company': ['Toyota', 'Honda', 'Nissan', 'Mitsubishi '], 'Price': [29995, 23600, 61500 , 58900]}
print(GermanCars)
print(japaneseCars)
```

```python
german_df = pd.DataFrame(GermanCars)
japanese_df = pd.DataFrame(japaneseCars)
cars = pd.concat([german_df,japanese_df], keys=['germany','japan'])
cars
```

Exercise 10: Merge two data frames using the following condition

```python
car_Price = {'Company': ['Toyota', 'Honda', 'BMV', 'Audi'], 'Price': [23845, 17995, 135925 , 71400]}
car_Horsepower = {'Company': ['Toyota', 'Honda', 'BMV', 'Audi'], 'horsepower': [141, 80, 182 , 160]}
```

```python
car_price = pd.DataFrame(car_Price)
car_hp = pd.DataFrame(car_Horsepower)
cars = pd.merge(car_price, car_hp, on='Company')
cars
```
