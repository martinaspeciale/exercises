Which was the most-ordered item?

```python
items = df.groupby('item_name')
quantities = items['quantity'].sum().sort_values(ascending = False)
print(f"quantities per item (top 10): \n{quantities.head(10)}")

print(f"\nmost-ordered item: {quantities.head(1)}\n")
```

For the most-ordered item, how many items were ordered?

```python
mostordered_item = quantities.index[0]
print(mostordered_item)

howmany_times = quantities.iloc[0]
print(howmany_times)
```

Turn the item price into a float

```python
dollarizer = lambda x : float(x[1:-1])
df['item_price'] = df['item_price'].apply(dollarizer)
```

```python
df.dtypes
```

```python
df.item_price.dtype
```

How much was the revenue for the period in the dataset?

```python
revenue = (df['quantity']*df['item_price']).sum()
print(f"Revenue was: $ {np.round(revenue,2)}")
```

Series.value_counts() counts the number of times each unique value appears in a pandas Series.

It returns a new Series where:

The index = unique values

The values = their counts

```python
print(f"df.order_id.unique() = {df.order_id.unique()}")
print(f"df.order_id.nunique() = {df.order_id.nunique()}")
print(f"df['order_id'].value_counts().count() = {df.order_id.value_counts().count()}")
```
