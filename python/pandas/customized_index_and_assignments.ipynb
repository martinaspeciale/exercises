```python
import pandas as pd
import numpy as np 
from faker import Faker 

np.random.seed(42)
fake = Faker()

n = 10 

data = {
    'Name' : [f"{fake.first_name()} {fake.last_name()}" for _ in range(n)],
    'Age'  : np.random.randint(20,60, size=n), 
    'City' : [fake.city() for _ in range(n)],
    'Score' : np.random.randint(0, 101, n)
}

df = pd.DataFrame(data, index=[f"P{i+1}" for i in range(n)])
df
```

```python
print(df.loc['P3'])
print(df.iloc[2])

df_filtered = df[(df['Age'] >= 30) & (df['Score'] >= 50)].copy()
df_filtered['Passed'] = df_filtered['Score'] >= 60
df_filtered
```
